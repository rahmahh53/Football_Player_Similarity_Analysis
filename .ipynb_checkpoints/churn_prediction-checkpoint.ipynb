**Risk Ratio**

```python
churn_no_partner/global_churn  # greater than 1, more likely to churn, 22% higher (1.*22*16..)
```

churn_partner/global_churn   # less than 1, less likely to churn, 25% lower 

IN SQL, select gender, avg(churn), avg(churn) - global_churn as diff, avg(churn)/global_churn as risk from data group by gender;
IN PANDAS:

```python
df_group = df_full_train.groupby('gender').churn.agg(['mean', 'count'])
df_group['diff'] = df_group['mean'] - global_churn
df_group['risk'] = df_group['mean'] / global_churn
df_group
```

```python
from IPython.display import display
```

```python
for c in categorical:
    print(c)
    df_group = df_full_train.groupby(c).churn.agg(['mean', 'count'])
    df_group['diff'] = df_group['mean'] - global_churn
    df_group['risk'] = df_group['mean'] / global_churn
    display(df_group)
```

### Feature Importance: Mutual Information

Mutual information is a measure of mutual dependence between two variables. It tells us how much we can learn about one variable if we know the value of another

```python
from sklearn.metrics import mutual_info_score
```

```python
mutual_info_score(df_full_train.churn, df_full_train.contract)
```

### Training Logistic Regression Model with Scikit-Learn

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
model.intercept_[0]
```

```python
model.coef_[0].round(3)
```

```python
model.predict_proba(X_train)  # soft predictions, not hard predictions which are just plain 0 and 1 
# and can be gotten with model.predict(X_train). For logistic regression, we want the probabilities.
```

```python
y_pred = model.predict_proba(X_val)[:, 1]
```

```python
churn_decision = (y_pred >= 0.5)
```

```python
df_val[churn_decision].customerid
```

```python
(y_val == churn_decision).mean()
```

```python
df_pred = pd.DataFrame()
df_pred['probability'] = y_pred
df_pred['prediction'] = churn_decision.astype(int)
df_pred['actual'] = y_val
```

```python
df_pred["correct"] = df_pred.prediction == df_pred.actual
```

```python
df_pred
```
