```python
import pandas as pd
import json
import numpy as np
```

```python
with open("../data/raw_data/competitions.json", "r") as f:
    data = json.load(f)
data[0]
```

```python
comp_list = []
for i in range(len(data)):
    comp_dict = {}
    comp_dict["competition_id"] = data[i]["competition_id"]
    comp_dict["competition_name"] = data[i]["competition_name"]
    comp_dict["country_name"] = data[i]["country_name"]
    comp_list.append(comp_dict)
comp_list[:10]
```

```python
comp_data = pd.DataFrame(comp_list)
comp_data.head()
```

```python
comp_data["competition_name"] = comp_data["competition_name"].str.replace(r"^\d+\.\s*", "", regex=True)
comp_data.head()
```

```python
seasons_list = []
for i in range(len(data)):
    seasons_dict = {}
    seasons_dict["season_id"] = data[i]["season_id"]
    seasons_dict["season_name"] = data[i]["season_name"]
    seasons_dict["competition_id"] = data[i]["competition_id"]
    seasons_list.append(seasons_dict)
seasons_list[:10]
```

```python
seasons_data = pd.DataFrame(seasons_list)
seasons_data.head()
```

```python
seasons_data = seasons_data.drop_duplicates(subset=["season_id"])
comp_data = comp_data.drop_duplicates(subset=["competition_id"])
seasons_data.head()
```
